# Exploration — France Travail API functions

Exploration only (see README: notebooks are not load-bearing). Goal: see exactly what `get_access_token` and `search_offers` return, using a single job offer, before relying on them in the collection script.

```python
from jobscope.scraping.france_travail import get_access_token, search_offers
```

## 1. Authentication

We only print the token's length — never the token itself, since it's a live credential.

```python
token = get_access_token()
print(f"token received, length={len(token)}")
```

## 2. Fetch a single offer

`range_start=0, range_end=0` is an inclusive range of size 1 — the API returns exactly one result.

```python
offers = search_offers(token, keywords="developpeur", range_start=0, range_end=0)
len(offers)
```

```python
offer = offers[0]
offer
```

## 3. Inspect the shape of the data

Which top-level fields does a raw offer actually carry? This is what will eventually need mapping onto the `JobPosting` schema.

```python
sorted(offer.keys())
```

```python
print("id:", offer.get("id"))
print("intitule:", offer.get("intitule"))
print("typeContrat:", offer.get("typeContrat"))
print("experienceLibelle:", offer.get("experienceLibelle"))
print("lieuTravail:", offer.get("lieuTravail"))
print("salaire:", offer.get("salaire"))
print("description (first 300 chars):", (offer.get("description") or "")[:300])
```
